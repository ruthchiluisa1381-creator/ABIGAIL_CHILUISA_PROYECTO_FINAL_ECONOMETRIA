```python
# ==========================================
# 01. ANÁLISIS EXPLORATORIO DE DATOS (EDA)
# ==========================================

import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Configuración de estilo visual
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# 1. CARGA DE DATOS PROCESADOS
DATA_PATH = os.path.join("..", "data", "processed", "enemdu_procesado.csv")

if os.path.exists(DATA_PATH):
    df = pd.read_csv(DATA_PATH, sep=';', low_memory=False)
    print(f"✅ Datos cargados correctamente. Filas: {df.shape[0]}, Columnas: {df.shape[1]}")
else:
    print(f"⚠️ No se encontró el archivo en {DATA_PATH}. Verifica la ejecución de 'limpiar_datos.py'.")

# 2. ESTADÍSTICAS DESCRIPTIVAS (CON Y SIN FACTOR DE EXPANSIÓN)
print("\n--- Estadísticas Sin Factor de Expansión ---")
print(df.describe().T[['mean', 'std', 'min', '50%', 'max']])

# Ejemplo con Factor de Expansión (asumiendo columna 'fexp' o similar si existe)
if 'fexp' in df.columns:
    print("\n--- Media Ponderada con Factor de Expansión ---")
    numeric_cols = df.select_dtypes(include=[np.number]).columns.drop('fexp', errors='ignore')
    weighted_means = {}
    for col in numeric_cols:
        valid = df[[col, 'fexp']].dropna()
        weighted_means[col] = (valid[col] * valid['fexp']).sum() / valid['fexp'].sum()
    print(pd.Series(weighted_means))

# 3. VISUALIZACIONES (Histogramas, Diagramas de Caja y Distribuciones)
# Ajusta 'ingreso', 'sexo' y 'area' según los nombres exactos de tus columnas
col_ingreso = 'ingreso' if 'ingreso' in df.columns else df.select_dtypes(include=[np.number]).columns[0]
col_sexo = 'sexo' if 'sexo' in df.columns else None
col_zona = 'area' if 'area' in df.columns else None

# Histograma de Ingresos / Variable Continua
plt.figure(figsize=(8, 5))
sns.histplot(df[col_ingreso].dropna(), kde=True, bins=30, color='skyblue')
plt.title(f'Distribución de {col_ingreso}')
plt.xlabel(col_ingreso)
plt.ylabel('Frecuencia')
plt.savefig(os.path.join("..", "outputs", "figures", "histograma_ingreso.png"), bbox_inches='tight')
plt.show()

# Diagrama de Caja por Género / Zona
if col_sexo and col_ingreso:
    plt.figure(figsize=(8, 5))
    sns.boxplot(x=col_sexo, y=col_ingreso, data=df, palette="Set2")
    plt.title(f'Distribución de {col_ingreso} por Género/Sexo')
    plt.savefig(os.path.join("..", "outputs", "figures", "boxplot_sexo.png"), bbox_inches='tight')
    plt.show()

if col_zona and col_ingreso:
    plt.figure(figsize=(8, 5))
    sns.boxplot(x=col_zona, y=col_ingreso, data=df, palette="Pastel1")
    plt.title(f'Distribución de {col_ingreso} por Zona/Área')
    plt.savefig(os.path.join("..", "outputs", "figures", "boxplot_zona.png"), bbox_inches='tight')
    plt.show()
```
